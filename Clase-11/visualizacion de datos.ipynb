## Visualizacion de datos: MatplotLib - Seaborn


```python
#https://colab.research.google.com/drive/1O3V439X-2rp7NpEXM2-Jjd9UlexKiLjk

import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

# Cambiar el tamaño del gráfico
plt.figure(figsize=(8, 5))

# Crear el gráfico con paleta diferente
sns.barplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="day",
    palette=sns.color_palette("magma", 4),
    legend=False
)

# Cambiar título y nombres de ejes
plt.title("Promedio de cuenta por día")
plt.xlabel("Día de la semana")
plt.ylabel("Cuenta total promedio")

# Agregar cuadrícula
plt.grid(axis="y")

plt.show()
```

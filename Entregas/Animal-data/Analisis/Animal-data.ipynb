#### Eliminamos columna vacia

```python
df = df.drop(columns="Animal code")

print(df.shape)
print("Animal code sigue presente?:", "Animal_code" in df.columns)
```

Procedemos a hacer una evaluacion de las columnas con informacion tipo "Str" para ello se crea una lista con los nombres de las columnas y un ciclo para ir por cada una y contar los valores unicos en estas

```python
columnas_categoricas = [
    "Animal name",
    "Animal type",
    "Gender",
    "Country",
    "Data compiled by"
    #Observation date 
    ]

for columna in columnas_categoricas:
    print(f"\n{'=' * 15} {columna} {'=' * 15}")
    print(df[columna].value_counts(dropna=False))
```

```python
df[columnas_categoricas] = df[columnas_categoricas].fillna("unknown")
print(df[columnas_categoricas])
```

```python
df["Gender"].value_counts()
```

```python
mapa_genero = {
    "male": "Male",
    "female": "Female",
    "unknown": "Unknown",
    "not determined": "Unknown"
}

df["Gender"] = df["Gender"].replace(mapa_genero)


print(df["Gender"].value_counts(dropna=False))
print("\nFaltantes en gender:", df["Gender"].isna().sum())
```

Ahora pasamos a otro requisito: convertir tipos de datos. Observation date seguramente fue leída como texto (object), pero debe convertirse a fecha para poder analizarla correctamente.

```python
print("Tipo actual:", df["Observation date"].dtype)

print("\nEjemplos de valores:")
print(df["Observation date"].head(10))
```

03.01.2024 usa el formato día.mes.año:  %d.%m.%Y

```python
df["Observation date"] = pd.to_datetime(
    df["Observation date"],
    format="%d.%m.%Y",
    errors="coerce"
)

print("Nuevo tipo:", df["Observation date"].dtype)
print("\nFechas inválidas o faltantes:", df["Observation date"].isna().sum())
print("\nPrimeras fechas convertidas:")
print(df["Observation date"].head())
```

```python
fechas_invalidas = df_raw.loc[
    df["Observation date"].isna(),
    "Observation date"
]

print("Cantidad de fechas inválidas:", fechas_invalidas.shape[0])
print("\nValores originales problemáticos:")
print(fechas_invalidas.value_counts(dropna=False))
```

como hay formatos mezclados, debemos volver a construir la columna desde df_raw, no desde df, porque en df esos 43 valores ya fueron convertidos a NaT

```python
df["Observation date"] = pd.to_datetime(
    df_raw["Observation date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print("Tipo:", df["Observation date"].dtype)
print("Fechas faltantes o inválidas:", df["Observation date"].isna().sum())
```

La variable Observation date estaba almacenada como texto y presentaba formatos inconsistentes: día.mes.año, mes.día.año y fechas con meses escritos en inglés. Después de la transformación, la columna quedó sin valores faltantes.

### Estandarizacion de animal type

```python
#Realizamos un diccionario con mapeo de variables
mapa_animal_type = {
    "red squireel": "Red squirrel",
    "red squirrel": "Red squirrel",
    "red squirel": "Red squirrel",

    "hedgehog": "Hedgehog",
    "ledgehod": "Hedgehog",
    "wedgehod": "Hedgehog",

    "lynx": "Lynx",
    "lynx?": "Lynx",

    "European bison": "European bison",
    "European bison™": "European bison",
    "European bisson": "European bison",
    "European buster": "European bison",

    "Unknown": "Unknown"
}

#Seleccionamos columna y actualizamos valores con .replace()
df["Animal type"] = df["Animal type"].replace(mapa_animal_type)

#Validamos ajustes
print(df["Animal type"].value_counts(dropna=False))
print("\nFaltantes en Animal type:", df["Animal type"].isna().sum())
print("\nNúmero de categorías:", df["Animal type"].nunique(dropna=False))
```

Basado en el analizis inicial, tiene sentido estadisitico analizar un dato como "Weight kg" para entender sus valores superiores a 10kg o menos

```python
weight = df["Weight kg"]
weight

df["weight category"] = np.where(
    weight > 10,
    "Above 10 kg",
    "10 kg or less"
)

df[["Weight kg", "weight category"]].head(20)
```

```python
mean = np.mean(weight)
print(f"El peso promedio de los animales en la base de datos es: ",  int(mean), "kg")
```

```python
median = np.median(weight)
print(f" La mediana estadistica del peso es: ", median )
```

### **Análisis con groupby()**

 Cuál es el número de observaciones por tipo de animal?

```python
cantidad_de_animales_por_tipo = df.groupby("Animal type").size() 
print(cantidad_de_animales_por_tipo)
```

Cuál es el promedio de una variable numérica según el tipo de animal?

```python
promedio = cantidad_de_animales_por_tipo.mean()
print(promedio)
```

Cuál es el promedio de una variable numérica según el país?

```python
AVG_Animal_type_segun_pais = df.groupby("Country")["Animal type"].count().mean()

print(AVG_Animal_type_segun_pais)
```
